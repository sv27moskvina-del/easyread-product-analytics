Кейсы
Вы — аналитик продукта EasyRead, популярного приложения для чтения книг. EasyRead позволяет пользователям читать книги онлайн, слушать аудиокниги и находить рекомендации по книгам на основе их интересов.

Команда продукта хочет улучшить ключевую метрику — конверсию в подписчиков (платящих клиентов). На текущий момент:

Всего у приложения 12 млн пользователей, из них 3 млн — платящие подписчики (конверсия: 25%).<br>
Для упрощения в этом задании будем считать, что количество новых привлеченных пользователей равно количеству ушедших пользователей.<br>
Продакт-менеджер сформировал 5 гипотез, которые могут улучшить конверсию. Он предоставил вводные данные о каждой гипотезе:<br>

Результаты предварительных исследований.<br>
Примерное время и ресурсы для реализации.<br>
Ваша задача — рассчитать, какая гипотеза имеет наивысший приоритет, используя методику ICE.<br>

Список гипотез:<br>
<b>Гипотеза 1. Рекомендации</b><br>
​​Если внедрим блок персонализированных рекомендаций, то сегменту подростков (20% неплатящих пользователей) не придется самостоятельно искать интересные книги, вследствие чего конверсия в подписчиков увеличится.

Impact: Ожидается, что 30% из сегмента подростков начнут конвертироваться в подписчиков.<br>
Confidence: фича с рекомендациями есть у всех конкурентов.<br>
Ease: Реализация требует работы 2 разработчиков и 1 аналитика в течение 2 недель.


<b>Гипотеза 2. Ускорение загрузки</b><br>
Если ускорим загрузку страниц книг, то пользователям с медленным интернетом не придется ждать, пока данные загрузятся, вследствие чего конверсия в подписчиков увеличится.

Impact: Изменения затронут 30% всех пользователей, увеличив их конверсию на 2 п.п.<br>
Confidence: среди клиентов есть запрос на этот функционал, у 1 ключевого конкурента есть такая фича, однако никаких исследований в рамках нашей компании - нет.<br>
Ease: Требуется 3 недели работы команды backend-разработчиков.


<b>Гипотеза 3. Тексты в email-рассылках</b><br>
Если внедрим более персонализированные тексты в email-рассылках, то пользователи будут меньше игнорировать письма, вследствие чего метрика конверсии в подписчиков увеличится.

Impact: Конверсия увеличится на 5% среди всех пользователей (подразумевается относительное увеличение).<br>
Confidence: A/B-тесты показали, что персонализация увеличивает открытие писем на 10%, но прирост подписок не тестировался.<br>
Ease: Требуется 3 дня работы копирайтера.


<b>Гипотеза 4. Пробный период с расширенным функционалом</b><br>
Если внедрим пробный период с расширенным функционалом для новых пользователей, то они смогут лучше оценить ценность сервиса, вследствие чего количество конверсий в платящих пользователей увеличится.

Impact: Пробный период с расширенным функционалом увеличит конверсию в платящих на 10% (подразумевается относительное увеличение).<br>
Confidence: Делали запуск крупномасштабного MVP, данные после запуска подтверждали гипотезу.<br>
Ease: Для реализации потребуется 1 неделя на доработку системы и настройки для разных тарифных планов.<br><br>
<b>Гипотеза 5. Тариф по сниженной цене</b><br>
Если введём тариф с базовыми функциями по сниженной цене, то пользователи с низким бюджетом смогут позволить себе подписку, вследствие чего общее количество подписчиков увеличится.

Impact: Привлечёт 5% неплатящих пользователей.<br>
Confidence: Продакт верит в эту гипотезу и внешний эксперт подтвердил, что гипотеза должна сработать.<br>
Ease: Потребуется 1 месяц работы всех команд: маркетинг, аналитика, разработка.<br>

```python
total_users = 12_000_000
paying_users = 3_000_000
non_paying_users = total_users - paying_users

impact_hypothesis_1 = non_paying_users * 0.2 * 0.3
print(f"Impact гипотезы 1: {impact_hypothesis_1:.0f} новых платящих пользователей")
```

```python
# Переведем impact в шкалу от 1 до 10 по формуле
# x_scaled = a + (x - x_min) * ((b - a) / (x_max - x_min))

# Задаём границы шкалы
a = 1
b = 10

# Исходные данные
mas_x = [540000, 72000, 150000, 300000, 450000]

x_min = min(mas_x)  # 72000
x_max = max(mas_x)  # 540000

# Масштабируем
for i in mas_x:
    x_scaled = a + (i - x_min) * ((b - a) / (x_max - x_min))
    print(f"Impact: {i} -> {x_scaled:.0f}")


```

```python
# Impact
impact = [10, 1, 2, 5, 8]

# Confidence оцениваем по условию
confidence = [4, 3, 7, 9, 2]

# Easy оцениваем по условию
easy = [5, 4, 9, 7, 2]

# считаем ICE
for i in range(len(impact)):
    ice = impact[i] * confidence[i] * easy[i]
    print(f"Гипотеза {i+1}: ICE = {ice}")
```
